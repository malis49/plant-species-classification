Архитектура модели

В проекте используется модель ResNet18, предварительно обученная на ImageNet.

Последний слой модели был заменён на линейный слой с 12 выходными нейронами.

Применялся подход transfer learning.

Визуальный анализ показывает, что некоторые классы имеют схожие морфологические признаки, что может усложнять задачу классификации.

```python
def show_images_from_class(class_name, n=3):
    class_path = os.path.join(data_dir, class_name)
    images = os.listdir(class_path)[:n]
    
    plt.figure(figsize=(10,3))
    for i, img_name in enumerate(images):
        img = Image.open(os.path.join(class_path, img_name))
        plt.subplot(1, n, i+1)
        plt.imshow(img)
        plt.axis("off")
    plt.suptitle(class_name)
    plt.show()

for cls in list(class_counts.keys())[:3]:
    show_images_from_class(cls)
```

Распределение классов является относительно сбалансированным, однако в отдельных классах наблюдается меньшее количество примеров, что может повлиять на точность модели.

```python
plt.figure(figsize=(10,6))
plt.bar(class_counts.keys(), class_counts.values())
plt.xticks(rotation=90)
plt.title("Распределение изображений по классам (train)")
plt.xlabel("Класс")
plt.ylabel("Количество изображений")
plt.tight_layout()
plt.show()
```

```python
data_dir = "../data/train"

class_counts = {}

for class_name in os.listdir(data_dir):
    class_path = os.path.join(data_dir, class_name)
    if os.path.isdir(class_path):
        class_counts[class_name] = len(os.listdir(class_path))

class_counts
```
